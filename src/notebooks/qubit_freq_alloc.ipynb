```python

from corral_crowding.allocation_optimizer import GateFidelityOptimizer
from corral_crowding.module_graph import QuantumModuleGraph
```

```python

n_4_edge_list = [
    ("Q0", "Q1", "qubit-qubit", "blue"),
    
    ("Q0", "Q3", "qubut-qubit", "blue"),
    ("Q1", "Q3", "qubit-qubit", "blue"),
    ("Q1", "Q2", "qubit-qubit", "blue"),
    ("Q2", "Q3", "qubit-qubit", "blue"),
    ("Q0", "SNAIL", "snail-qubit", "orange"),
    ("Q1", "SNAIL", "snail-qubit", "orange"),
    ("Q2", "SNAIL", "snail-qubit", "orange"),
    ("Q3", "SNAIL", "snail-qubit", "orange"),
]

qmg_4 = QuantumModuleGraph(num_qubits=4, edges=n_4_edge_list)


optimizer = GateFidelityOptimizer(
    qmg_4,
    min_bare_space_ghz=0.12,
    lambdaq=0.08,
    eta=1.8,
    g3=60e6,
    qubit_bounds=(3.3, 5.7),
    snail_bounds=(4.2, 4.7),
    use_lifetime=False,
)

optimizer.optimize_frequencies(500)
optimizer.report_results()

#[5.65797416 3.38890601 5.53793509 5.41787659], 4.433098459148965
```
