```python
from corral_crowding.qutrit_module_graph import QutritModuleGraph
from corral_crowding.qutrit_allocation_optimizer import GateFidelityOptimizer
```

```python
n = 3
base_freq = [5.0, 5.5, 6.0]
anharmonicities = [-0.29, -0.31, -0.3]
qmg:QutritModuleGraph = QutritModuleGraph(n, edge_type="all-to-all")
```

```python
qmg.plot_interaction_frequencies(qutrit_frequencies=base_freq, qutrit_anharmonicities=anharmonicities, snail_frequency=6.7)
qmg.plot_graph(base_freq, snail_frequency=6.7)
```

```python

optimizer = GateFidelityOptimizer(
    qmg,
    min_bare_space_ghz=0.12,
    lambdaq=0.08,
    eta=1.8,
    g3=60e6,
    qubit_bounds=(3.3, 5.7),
    snail_bounds=(4.2, 4.7),
    use_lifetime=True,
)

optimizer.optimize_frequencies(256)
optimizer.report_results()
```
